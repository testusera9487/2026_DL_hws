# 作業 2：批次輸入的多隱藏層全連接神經網路

- 輸入層：6 個神經元
- 隱藏層 1：5 個神經元，使用 ReLU
- 隱藏層 2：4 個神經元，使用 ReLU
- 隱藏層 3：4 個神經元，使用 Sigmoid
- 輸出層：3 個神經元，使用 Softmax


## 1. 批次資料與網路架構

定義 5 個樣本，每個樣本有 6 個特徵值，並且每個樣本對應一個 3 類別的 one-hot 標籤。
- X shape: (5, 6)
- Y shape: (5, 3)
- W1: (6, 5), b1: (5,)
- W2: (5, 4), b2: (4,)
- W3: (4, 4), b3: (4,)
- W4: (4, 3), b4: (3,)

```python
import numpy as np
np.set_printoptions(precision=6, suppress=True)

X = np.array([
    [0.5, -0.2, 1.3, 0.8, -1.0, 0.4],
    [1.2, 0.7, -0.3, 1.5, 0.2, -0.4],
    [-0.8, 1.1, 0.6, -0.5, 0.9, 0.3],
    [0.1, -1.4, 0.9, 2.0, 0.6, -0.1],
    [-0.3, 0.8, 1.6, -0.7, 0.4, 1.2],
], dtype=float)

Y = np.array([
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1],
    [1, 0, 0],
    [0, 1, 0],
], dtype=float)

print('X shape =', X.shape)
print(X)
print('Y shape =', Y.shape)
print(Y)
```

## 2. 定義函式

```python
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def softmax(x):
    x = x - np.max(x, axis=1, keepdims=True)
    e_x = np.exp(x)
    return e_x / np.sum(e_x, axis=1, keepdims=True)

def cross_entropy_error(y, y_hat):
    eps = 1e-12
    y = np.asarray(y, dtype=float)
    y_hat = np.asarray(y_hat, dtype=float)
    return -np.sum(y * np.log(y_hat + eps)) / y.shape[0]

print('函數定義完成')
```

## 3. 權重與偏值設定

沿用與hw1 相同的權重

```python
W1 = np.array([
    [0.2, -0.1, 0.3, -0.2, 0.4],
    [-0.5, 0.6, 0.1, -0.1, 0.2],
    [0.1, -0.2, 0.4, 0.3, -0.3],
    [0.5, 0.2, -0.1, 0.6, -0.4],
    [-0.3, 0.4, 0.2, -0.5, 0.3],
    [0.7, -0.2, 0.5, 0.1, -0.3],
], dtype=float)
b1 = np.array([0.1, -0.2, 0.3, 0.2, -0.1], dtype=float)

W2 = np.array([
    [0.3, -0.2, 0.5, 0.1],
    [0.4, 0.2, -0.1, 0.3],
    [-0.3, 0.5, 0.2, -0.2],
    [0.1, -0.4, 0.6, 0.2],
    [0.5, 0.1, -0.2, 0.4],
], dtype=float)
b2 = np.array([0.2, -0.1, 0.4, 0.3], dtype=float)

W3 = np.array([
    [0.5, -0.2, 0.3, 0.1],
    [0.1, 0.4, -0.3, 0.2],
    [-0.1, 0.2, 0.5, -0.4],
    [0.3, -0.1, 0.2, 0.4],
], dtype=float)
b3 = np.array([-0.2, 0.1, 0.3, -0.1], dtype=float)

W4 = np.array([
    [0.2, -0.3, 0.5],
    [0.4, 0.1, -0.2],
    [-0.1, 0.3, 0.4],
    [0.5, -0.2, 0.1],
], dtype=float)
b4 = np.array([0.1, -0.2, 0.3], dtype=float)

print('W1 shape =', W1.shape, 'b1 shape =', b1.shape)
print(W1)
print('b1 =', b1)
print('W2 shape =', W2.shape, 'b2 shape =', b2.shape)
print(W2)
print('b2 =', b2)
print('W3 shape =', W3.shape, 'b3 shape =', b3.shape)
print(W3)
print('b3 =', b3)
print('W4 shape =', W4.shape, 'b4 shape =', b4.shape)
print(W4)
print('b4 =', b4)
```

## 4. 批次前向傳播

讓每一筆樣本都通過相同的多層神經網路，並保留批次維度。

```python
z1 = X @ W1 + b1
h1 = relu(z1)

z2 = h1 @ W2 + b2
h2 = relu(z2)

z3 = h2 @ W3 + b3
h3 = sigmoid(z3)

z4 = h3 @ W4 + b4
y_hat = softmax(z4)

print('z1 shape =', z1.shape)
print(z1)
print('h1 shape =', h1.shape)
print(h1)
print('z2 shape =', z2.shape)
print(z2)
print('h2 shape =', h2.shape)
print(h2)
print('z3 shape =', z3.shape)
print(z3)
print('h3 shape =', h3.shape)
print(h3)
print('z4 shape =', z4.shape)
print(z4)
print('y_hat shape =', y_hat.shape)
print(y_hat)
```

## 5. 批次交叉熵誤差

這裡利用批次版的 cross entropy error，先計算每筆資料的交叉熵，再對 5 筆平均。

```python
loss = cross_entropy_error(Y, y_hat)
print('batch cross_entropy_error =', loss)
```
